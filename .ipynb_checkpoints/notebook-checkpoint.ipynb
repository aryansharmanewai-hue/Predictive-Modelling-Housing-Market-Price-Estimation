#  House Price Pridiction

### Data preprocessing libraries

```python
import numpy as np
import pandas as pd
```

```python
# Replace with your dataset path
df = pd.read_csv("d:\MyData\Downloads\house_price_dataset.csv")

# View data
df.head(7)
```

```python
print(df.shape)
print(df.columns)
print(df.info())
print(df.describe())
print(df.dtypes)
```

```python
# Check missing values
df.isnull().sum()

# Fill missing values (basic approach)
df.fillna(df.mean(numeric_only=True), inplace=True)
```

### Data Visulaization

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Scatter plot
# Replace 'features' and 'price' with your column names
features = ['BHK','Area','House_Age','Location']

for feature in features:
    plt.figure()
    sns.scatterplot(x=df[feature], y=df['Price'])
    plt.title(f"{feature} vs Price")
    plt.show()
```

```python
# Bar plot
features = ['BHK','Area','House_Age','Location']

for feature in features:
    plt.figure(figsize=(10,6))
    sns.barplot(x=df[feature], y=df['Price'])
    plt.title(f"{feature} vs Price")
    plt.xticks(rotation=45)
    plt.show()
```

### Model Training

```python
# Feature Selection

X = df.drop("Price", axis=1)   # input features
y = df["Price"]                # target
```

```python
#df = pd.get_dummies(df, columns=['location'], drop_first=True)
```

```python
print(X.head())
```

```python
# Model Training
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# split the data so that model can testing on unseen data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2)
```

##  Linear Regression

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
lr_model = LinearRegression()
lr_model.fit(X_train, y_train)

y_pred_lr = lr_model.predict(X_test)
```

```python
# Evaluate linear regression model
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

print("MAE:", mean_absolute_error(y_test, y_pred_lr))
print("MSE:", mean_squared_error(y_test, y_pred_lr))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred_lr)))
print("R2 Score:", r2_score(y_test, y_pred_lr))
```

```python
# VISUALIZATION 
import matplotlib.pyplot as plt

plt.scatter(y_test, y_pred_lr)
plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Actual vs Predicted")
plt.show()
```

## Decision Tree

```python
#Model Traing
from sklearn.tree import DecisionTreeRegressor

dt = DecisionTreeRegressor(random_state=42)
dt.fit(X_train, y_train)
```

```python
#Prediction
y_pred_dt = dt.predict(X_test)
```

```python
#Model_Evaluation
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

print("MAE:", mean_absolute_error(y_test, y_pred_dt))
print("MSE:", mean_squared_error(y_test, y_pred_dt))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred_dt))
)
print("R2 Score:", r2_score(y_test, y_pred_dt))
```

```python
#Visulization
import matplotlib.pyplot as plt

plt.scatter(y_test, y_pred_dt)
plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Decision Tree Prediction")
plt.show()
```

## Random Forest

```python
#Train Model
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(
    n_estimators=100,   # number of trees
    random_state=42
)

rf.fit(X_train, y_train)
```

```python
#Prediction
y_pred_rf = rf.predict(X_test)
```

```python
#Model Evaluation
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

print("MAE:", mean_absolute_error(y_test, y_pred_rf))
print("MSE:", mean_squared_error(y_test, y_pred_rf))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred_rf)))
print("R2 Score:", r2_score(y_test, y_pred_rf))
```

```python
#Visulization
import matplotlib.pyplot as plt

plt.scatter(y_test, y_pred_rf)
plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Random Forest Prediction")
plt.show()
```
